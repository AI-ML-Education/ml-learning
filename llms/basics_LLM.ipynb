## LLM Architecture
Lets first look into what a LLM is made of. A LLM is made up of several components that work together to process and generate human-like text. Here are the key components:
- **Tokenizer**:  
This component breaks down the input text into smaller units called tokens. Tokens can be words, subwords, or characters, depending on the tokenizer used. The tokenizer also converts these tokens into numerical representations (token IDs) that the model can understand.

- **Embedding Layer**:  
This layer converts the token IDs into dense vectors of fixed size. These vectors capture semantic information about the tokens and their relationships to each other.

- **Transformer Layers**:  
The core of a LLM is made up of multiple transformer layers. Each layer consists of:
   - **Multi-Head Self-Attention Mechanism**:  
   This mechanism allows the model to focus on different parts of the input text when generating each token. It helps the model understand context and relationships between words.
   
   - **Feed-Forward Neural Network**:  
   This is a fully connected neural network that processes the output of the attention mechanism. It applies non-linear transformations to capture complex patterns in the data.
   
   - **Layer Normalization and Residual Connections**:  
   These techniques help stabilize training and improve the flow of gradients through the network.

- **Output Layer**:  
The final layer of the model is a linear layer followed by a softmax function. This layer generates a probability distribution over the vocabulary for the next token in the sequence.

- **Training Objective**:  
LLMs are typically trained using a language modeling objective, such as predicting the next token in a sequence (causal language modeling) or filling in missing tokens (masked language modeling). The model learns to minimize the difference between its predictions and the actual tokens in the training data.

- **Fine-tuning**:  
After pre-training on a large corpus of text, LLMs can be fine-tuned on specific tasks or domains to improve their performance in those areas.

- **Inference**:  
During inference, the model generates text by sampling tokens from the probability distribution produced by the output layer. Various decoding strategies, such as greedy decoding, beam search, or sampling, can be used to generate coherent and contextually relevant text.
